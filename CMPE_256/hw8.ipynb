```python
import networkx as nx
from community import community_louvain as co
```

```python
G = nx.DiGraph()

with open('wiki-Vote.txt', 'r') as f:
    for line in f:
        if line.startswith('#'):
            continue
        u, v = map(int, line.strip().split())
        G.add_edge(u, v)
```

```python
print(f"Number of nodes: {G.number_of_nodes()}")
print(f"Number of edges: {G.number_of_edges()}")
```

1: Number of nodes and number of edges [1pt]

There are 7115 nodes and 103689 edges in the graph.

```python
is_strongly_connected = nx.is_strongly_connected(G)
print(f"The graph is {is_strongly_connected} connected")

if is_strongly_connected:
    num_components = nx.number_of_strongly_connected_components(G)
    print(f"{num_components} connected components")
else:
    num_components = nx.number_weakly_connected_components(G)
    print(f"{num_components} connected components")
```

2: Is the graph strongly connected? If yes, return the number of strongly connected components. If not, return the number of weakly connected components.  [1pt]

The graph is a weakly connected graph with 24 weak connections

```python
longest_shortest_path = 0
for component_nodes in nx.weakly_connected_components(G):
    component_subgraph = G.subgraph(component_nodes)
    undirected_component = component_subgraph.to_undirected()

    if nx.is_connected(undirected_component) and len(undirected_component) > 1:
          component_diameter = nx.diameter(undirected_component)
          longest_shortest_path = max(longest_shortest_path, component_diameter)
    elif len(undirected_component) == 1:
        pass

longest_shortest_path
```
